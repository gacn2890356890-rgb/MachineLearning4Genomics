## 🤖 模型 (Model)
所有的魔法模型都由 3 个核心部件组成哦：
- **任务 (Task)**: 我们究竟想让模型解决什么问题？
- **学习方法 (Learning methodology)**: ML 算法可以是监督的，也可以是无监督的。这决定了它的学习方式。
- **性能指标 (Performance measure)**: 我们用来评估模型表现的量化标准。

### 1. 任务 (Task)
我们想让模型解决什么问题呢？
- **定义 f(x)**: 什么样的函数能把我们的 x (输入) 映射得尽可能接近真实的 Y (输出) 呢？
    - f(x) 的终极目标就是预测出一个尽可能接近真实 Y 的 Y*。
    - 给定解释性特征后，模型 (`f(x) + e = Y*`) 就会推导出预测结果 (Y)。

这里的 **e** 是*不可约误差 (irreducible error)*。它包含了测量误差、数据随机性以及模型选择不当等因素造成的偏差。不管你怎么拼命优化模型，这个 e 永远都不可能变成 0 哦！(这就是现实的残酷呢 😢)

- **特征工程与特征选择**: x 到底是什么？我们要怎么决定把哪些解释性特征塞进模型里呢？
- **我们的 f(x) 对这个问题来说是正确的吗？**: 模型对数据做了什么假设？我们需要提前给数据“变装”（转换数据）吗？

建立模型，其实就是把你的研究问题翻译成“机器学习的语言”。一个机器学习任务通常包含解释性特征 (explanatory features) 和一个结果特征 (outcome feature)。

**结果特征** 就是当解释性特征发生变化时，我们预期会随之改变的那个特征。 *(实验室备忘：这里以后要补充一个机器学习的小例子呢)*


### 2. 学习方法 (Learning Methodology)
我们的模型是监督的还是无监督的？这会怎么影响它的学习过程呢？
- 我们的 **ML 模型是怎么学习的？**: 概览一下模型是如何“自学成才”的。
    - 算法的学习方式取决于你喂给它什么类型的数据。
    - 有标签的数据会使用“监督学习”，而没贴标签的数据只能用“无监督学习”啦。
    - 模型毕生的追求就是：**最小化损失函数**！
- 什么是我们的 **损失函数 (Loss function)** 呢？: 每一个监督模型都有一个它做梦都想消除的损失函数。
    - 损失函数量化了这么一种感觉：当正确输出明明是 y 时，你用 f(x) 预测出了个 Y*，此时你会有多“不开心”。😣
    - 它就是我们要极力缩小的大魔王！
    - 损失函数也衡量了我们的 f(x) 与数据的契合度。
    - 选择哪种损失函数，完全取决于你的任务类型哦。
 
 **回归 (Regression)**: [绝对误差 (L1) 和 最小二乘误差 (L2)](http://www.chioka.in/differences-between-l1-and-l2-as-loss-function-and-regularization/)，这还包括了均方误差 (MSE) 和均方根误差 (RMSE)。
 
 **分类 (Classification)**: 对数损失 (log loss) 和 铰链损失 (hinge loss)。
 
- **优化过程 (Optimization process)**: 模型到底是怎么去缩小这个损失函数的呢？

这种不断调整 f(x) 来降低损失函数的过程，就被称为“学习”。正是这个过程，让普通的最小二乘 (OLS) 回归摇身一变，成为了高大上的机器学习算法呢！

### 3. 性能指标 (Performance measure)
- **性能衡量尺**: R2、调整后的 R2、MSE
- **特征表现**: 统计显著性、p 值 
- **应对未知数据的泛化能力**: 过拟合 (Overfitting)、欠拟合 (underfitting)、偏差 (bias)、方差 (variance)
